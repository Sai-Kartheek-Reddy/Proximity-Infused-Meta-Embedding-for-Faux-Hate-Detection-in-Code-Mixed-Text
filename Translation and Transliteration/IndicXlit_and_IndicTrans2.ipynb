```python
from indicTrans.inference.engine import Model

indic2en_model = Model(expdir='../indic-en')
```

```python

# Function to transliterate Telugu text to English
def translate_to_english(text):
    english_text = indic2en_model.translate_paragraph(text, 'te', 'en') # hi for hindi, te for telugu
    return english_text
```

```python
import pandas as pd
from tqdm import tqdm
df = pd.read_csv('<Dataset_Path>')
```

```python
df['<New_Column_Name>']=None # Here we are creating the Empty column to store the Translated Text
for n, text in enumerate(tqdm(df['<Column_Name>'])): # Here we are giving the input Column Header Name
  df['New_Column_Name'][n] = translate_to_english(text)
  # break
df.head()
```
